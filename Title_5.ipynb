## Binary Search + Maximum and Minimum

### QUESTION 1: SEARCH AN ELEMENT IN A SORTED ARRAY

```python
def binary_search_q1(arr, key):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == key:
            print("Element found at position", mid + 1)
            return

        elif arr[mid] < key:
            low = mid + 1

        else:
            high = mid - 1

    print("Element not found")


print("\n========== QUESTION 1 - TEST CASE 1 ==========")
binary_search_q1([10, 20, 30, 40, 50], 30)

print("\n========== QUESTION 1 - TEST CASE 2 ==========")
binary_search_q1([10, 20, 30, 40, 50], 25)
```

### QUESTION 2: FIND THE INDEX OF A GIVEN NUMBER

```python
def binary_search_q2(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid

        elif arr[mid] < target:
            low = mid + 1

        else:
            high = mid - 1

    return -1


print("\n========== QUESTION 2 - TEST CASE 1 ==========")
arr = [2, 4, 6, 8, 10, 12]
target = 8

index = binary_search_q2(arr, target)
print("Index =", index)


print("\n========== QUESTION 2 - TEST CASE 2 ==========")
arr = [2, 4, 6, 8, 10, 12]
target = 5

index = binary_search_q2(arr, target)
print("Index =", index)
```

### QUESTION 3: COUNT NUMBER OF ITERATIONS

```python
def binary_search_iterations(arr, key):
    low = 0
    high = len(arr) - 1
    iterations = 0

    while low <= high:

        iterations += 1

        mid = (low + high) // 2

        if arr[mid] == key:
            print("Element found")
            print("Iterations =", iterations)
            return

        elif arr[mid] < key:
            low = mid + 1

        else:
            high = mid - 1

    print("Element not found")
    print("Iterations =", iterations)


print("\n========== QUESTION 3 - TEST CASE 1 ==========")
binary_search_iterations(
    [5, 10, 15, 20, 25, 30, 35],
    25
)

print("\n========== QUESTION 3 - TEST CASE 2 ==========")
binary_search_iterations(
    [5, 10, 15, 20, 25, 30, 35],
    40
)
```

### QUESTION 4: FIRST OCCURRENCE

```python
def first_occurrence(arr, key):
    low = 0
    high = len(arr) - 1
    result = -1

    while low <= high:

        mid = (low + high) // 2

        if arr[mid] == key:
            result = mid
            high = mid - 1

        elif arr[mid] < key:
            low = mid + 1

        else:
            high = mid - 1

    return result


print("\n========== QUESTION 4 - TEST CASE 1 ==========")

arr = [1, 2, 2, 2, 3, 4, 5, 6]
key = 2

index = first_occurrence(arr, key)

if index != -1:
    print("First occurrence at index", index)
elif index == -1:
    print("Element not found")


print("\n========== QUESTION 4 - TEST CASE 2 ==========")

arr = [1, 2, 2, 2, 3, 4, 5, 6]
key = 7

index = first_occurrence(arr, key)

if index != -1:
    print("First occurrence at index", index)
elif index == -1:
    print("Element not found")
```

### QUESTION 5: LAST OCCURRENCE

```python
def last_occurrence(arr, key):
    low = 0
    high = len(arr) - 1
    result = -1

    while low <= high:

        mid = (low + high) // 2

        if arr[mid] == key:
            result = mid
            low = mid + 1

        elif arr[mid] < key:
            low = mid + 1

        else:
            high = mid - 1

    return result


print("\n========== QUESTION 5 - TEST CASE 1 ==========")

arr = [1, 2, 2, 2, 3, 4, 5, 6]
key = 2

index = last_occurrence(arr, key)

if index != -1:
    print("Last occurrence at index", index)
elif index == -1:
    print("Element not found")


print("\n========== QUESTION 5 - TEST CASE 2 ==========")

arr = [1, 2, 2, 2, 3, 4, 5, 6]
key = 9

index = last_occurrence(arr, key)

if index != -1:
    print("Last occurrence at index", index)
elif index == -1:
    print("Element not found")
```

## DIVIDE AND CONQUER MINIMUM AND MAXIMUM

```python
def find_min_max(arr, low, high):

    # Only one element
    if low == high:
        return arr[low], arr[low]

    # Two elements
    if high == low + 1:

        if arr[low] < arr[high]:
            return arr[low], arr[high]
        else:
            return arr[high], arr[low]

    # Divide
    mid = (low + high) // 2

    left_min, left_max = find_min_max(
        arr,
        low,
        mid
    )

    right_min, right_max = find_min_max(
        arr,
        mid + 1,
        high
    )

    # Conquer
    minimum = min(left_min, right_min)
    maximum = max(left_max, right_max)

    return minimum, maximum
```

### QUESTION 6: HIGHEST AND LOWEST TEMPERATURE

```python
print("\n========== QUESTION 6 - TEMPERATURE ==========")

temperatures = [32, 28, 35, 25, 31, 29, 37]

minimum, maximum = find_min_max(
    temperatures,
    0,
    len(temperatures) - 1
)

print("Minimum Temperature =", minimum)
print("Maximum Temperature =", maximum)
```
