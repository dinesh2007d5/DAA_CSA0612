## SEQUENTIAL SEARCH - QUESTION 6
### SENTINEL SEARCH

```python
def ordinary_search(arr, key):
    comparisons = 0

    for i in range(len(arr)):
        comparisons += 1
        if arr[i] == key:
            return i, comparisons

    return -1, comparisons

def sentinel_search(arr, key):
    a = arr.copy()
    n = len(a)

    a.append(key)

    i = 0
    comparisons = 0

    while True:
        comparisons += 1

        if a[i] == key:
            break

        i += 1

    if i < n:
        return i, comparisons
    else:
        return -1, comparisons
```

```python
def sequential_search_q6():
    arr = [14, 9, 22, 35, 18, 41, 27]
    key = 18

    pos1, comp1 = ordinary_search(arr, key)
    pos2, comp2 = sentinel_search(arr, key)

    print("Ordinary Sequential Search:")
    print("Position found =", pos1 + 1)
    print("Comparison count =", comp1)

    print("\nSentinel Sequential Search:")
    print("Position found =", pos2 + 1)
    print("Comparison count =", comp2)

print("\n--- Sequential Search Q6 ---")
sequential_search_q6()
```

## SEQUENTIAL SEARCH - QUESTION 10

```python
def sequential_search_q10(arr, key):
    comparisons = 0

    print("\nSearching for:", key)

    for i in range(len(arr)):
        comparisons += 1

        print(
            "Comparison", comparisons,
            ":", arr[i], "vs", key,
            "->",
            "Match" if arr[i] == key else "Mismatch"
        )

        if arr[i] == key:
            print("Found at position:", i + 1)
            print("Total comparisons:", comparisons)
            return comparisons

    print("Element not found")
    print("Total comparisons:", comparisons)

    return comparisons
```

```python
print("\n--- Sequential Search Q10 ---")

arr = [
    45, 23, 67, 12, 89,
    34, 56, 78, 90, 11,
    29, 73, 18, 64, 37
]

sequential_search_q10(arr, 73)
sequential_search_q10(arr, 18)
sequential_search_q10(arr, 100)

print("\nPerformance Analysis:")
print("Best Case    : O(1)")
print("Average Case : O(n)")
print("Worst Case   : O(n)")
print("Space        : O(1)")
```

## BRUTE FORCE STRING MATCHING - COMMON FUNCTION

```python
def brute_force_matching(text, pattern):
    positions = []
    comparisons = 0

    n = len(text)
    m = len(pattern)

    for i in range(n - m + 1):

        match = True

        for j in range(m):
            comparisons += 1

            if text[i + j] != pattern[j]:
                match = False
                break

        if match:
            positions.append(i)

    return positions, comparisons
```

## STRING MATCHING - QUESTION 1

```python
def string_q1():
    text = "AABAACAADAABAABA"
    pattern = "AABA"

    positions, comparisons = brute_force_matching(text, pattern)

    print("Pattern positions (0-based):", positions)
    print("Pattern positions (1-based):",
          [p + 1 for p in positions])
    print("Total number of comparisons:", comparisons)
```

```python
print("\n--- String Matching Q1 ---")
string_q1()
```

## STRING MATCHING - QUESTION 9

```python
def string_q9():
    text = "ABCDABCABCDA"
    pattern = "ABCDA"

    n = len(text)
    m = len(pattern)

    print("Alignment | Matching Result")
    print("-" * 35)

    occurrences = []

    for i in range(n - m + 1):

        match = True

        for j in range(m):
            if text[i + j] != pattern[j]:
                match = False
                break

        result = "Match" if match else "Mismatch"

        print(f"{i + 1:9} | {result}")

        if match:
            occurrences.append(i + 1)

    print("\nPattern occurrence positions:", occurrences)
```

## STRING MATCHING - QUESTION 10

```python
def string_q10():
    text = "TTATAGATCTCGTATTCTTTATAGATCTCCTATTCTT"
    pattern = "TATCTT"

    n = len(text)
    m = len(pattern)

    occurrences = []
    total_comparisons = 0

    print("TEXT   :", text)
    print("PATTERN:", pattern)

    print("\nShifting Process")
    print("-" * 70)

    for i in range(n - m + 1):

        comparisons = 0
        match = True

        for j in range(m):

            comparisons += 1
            total_comparisons += 1

            if text[i + j] != pattern[j]:
                match = False
                break

        aligned_pattern = " " * i + pattern

        print(
            f"Shift {i:2}: {aligned_pattern}"
            f"   -> {'MATCH' if match else 'MISMATCH'}"
            f" ({comparisons} comparisons)"
        )

        if match:
            occurrences.append(i + 1)

    print("\nResults")
    print("-" * 40)

    print("Pattern occurrences (1-based):", occurrences)
    print("Total comparisons:", total_comparisons)

    print("\nComplexity Analysis")
    print("Best Case   : O(n)")
    print("Worst Case  : O(n*m)")
    print("Space       : O(1)")
```

```python
print("\n--- String Matching Q10 ---")
string_q10()
```
