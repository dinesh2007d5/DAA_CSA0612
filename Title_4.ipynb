### TOPIC 1: CLOSEST PAIR OF POINTS

```python
def euclidean_distance(p1, p2):
    return sqrt((p1[0] - p2[0]) ** 2 +
                (p1[1] - p2[1]) ** 2)

def closest_pair(points):
    min_distance = float('inf')
    closest = None

    for i in range(len(points)):
        for j in range(i + 1, len(points)):
            distance = euclidean_distance(points[i], points[j])

            if distance < min_distance:
                min_distance = distance
                closest = (points[i], points[j])

    return closest, min_distance
```

### CONVEX HULL BASIC BRUTE FORCE (QUESTION 5 CONTINUED)

```python
def orientation(p, q, r):
    value = ((q[0] - p[0]) * (r[1] - p[1]) -
             (q[1] - p[1]) * (r[0] - p[0]))

    if value > 0:
        return 1       # Counter-clockwise

    if value < 0:
        return -1      # Clockwise

    return 0           # Collinear


def convex_hull_brute(points):
    n = len(points)

    if n <= 2:
        return points

    hull_edges = set()

    for i in range(n):
        for j in range(i + 1, n):

            left = False
            right = False

            for k in range(n):

                if k == i or k == j:
                    continue

                side = orientation(points[i], points[j], points[k])

                if side > 0:
                    left = True

                elif side < 0:
                    right = True

            # Edge belongs to hull if all other points
            # are on one side or collinear.
            if not (left and right):
                hull_edges.add((i, j))

    # Collect hull points
    hull_points = set()

    for i, j in hull_edges:
        hull_points.add(points[i])
        hull_points.add(points[j])

    # Arrange hull points around centroid
    center_x = sum(p[0] for p in hull_points) / len(hull_points)
    center_y = sum(p[1] for p in hull_points) / len(hull_points)

    def angle(p):
        import math
        return math.atan2(p[1] - center_y, p[0] - center_x)

    return sorted(hull_points, key=angle)


print("\n========== CONVEX HULL Q5 ==========")

points = [
    (10, 0),       # P1
    (11, 5),       # P2
    (5, 3),        # P3
    (9, 3.5),      # P4
    (15, 3),       # P5
    (12.5, 7),     # P6
    (6, 6.5),      # P7
    (7.5, 4.5)     # P8
]

hull = convex_hull_brute(points)

print("Convex Hull Points:")

for p in hull:
    print(p)

print("\nTime Complexity: O(n^3)")
print("Space Complexity: O(n)")
```

### HANDLING COLLINEAR POINTS

```python
def convex_hull_with_collinear(points):
    """
    Brute-force convex hull that removes interior collinear
    points and keeps only extreme boundary points.
    """

    n = len(points)

    if n <= 2:
        return points

    hull = set()

    for i in range(n):
        for j in range(i + 1, n):

            left = False
            right = False

            for k in range(n):
                if k == i or k == j:
                    continue

                side = orientation(points[i], points[j], points[k])

                if side > 0:
                    left = True
                elif side < 0:
                    right = True

            if not (left and right):
                hull.add(points[i])
                hull.add(points[j])

    center_x = sum(p[0] for p in hull) / len(hull)
    center_y = sum(p[1] for p in hull) / len(hull)

    import math

    hull = sorted(
        hull,
        key=lambda p: math.atan2(
            p[1] - center_y,
            p[0] - center_x
        )
    )

    return hull
```

### TOPIC 3: EXHAUSTIVE SEARCH FOR SUBSET SUM

```python
def subset_sum_exhaustive(numbers, target):
    n = len(numbers)

    for r in range(n + 1):

        for subset in combinations(numbers, r):

            if sum(subset) == target:
                return list(subset)

    return None
```

```python
print("\n========== SUBSET SUM Q11 ==========")

numbers = [3, 34, 4, 12, 5, 2]
target = 9

subset = subset_sum_exhaustive(numbers, target)

if subset:
    print("Subset found:", subset)
    print("Sum:", sum(subset))
else:
    print("No subset found")
```

#### QUESTION 12: EXHAUSTIVE SEARCH - TSP

```python
def calculate_path_distance(path):
    distance = 0

    for i in range(len(path) - 1):
        distance += euclidean_distance(path[i], path[i + 1])

    return distance


def tsp_exhaustive(cities):
    start = cities[0]

    best_distance = float('inf')
    best_path = None

    remaining = cities[1:]

    for perm in permutations(remaining):

        path = [start] + list(perm) + [start]

        distance = calculate_path_distance(path)

        if distance < best_distance:
            best_distance = distance
            best_path = path

    return best_path, best_distance


print("\n========== TSP Q12 - TEST CASE 1 ==========")

cities1 = [
    (1, 2),
    (4, 5),
    (7, 1),
    (3, 6)
]

path, distance = tsp_exhaustive(cities1)

print("Shortest Distance:", distance)
print("Shortest Path:", path)


print("\n========== TSP Q12 - TEST CASE 2 ==========")

cities2 = [
    (2, 4),
    (8, 1),
    (1, 7),
    (6, 3),
    (5, 9)
]

path, distance = tsp_exhaustive(cities2)

print("Shortest Distance:", distance)
print("Shortest Path:", path)
```

#### QUESTION 13: EXHAUSTIVE SEARCH - ASSIGNMENT PROBLEM

```python
def total_cost(assignment, cost_matrix):
    total = 0

    for worker, task in assignment:
        total += cost_matrix[worker][task]

    return total


def assignment_problem(cost_matrix):
    n = len(cost_matrix)

    best_cost = float('inf')
    best_assignment = None

    for task_permutation in permutations(range(n)):

        assignment = []

        for worker in range(n):
            task = task_permutation[worker]
            assignment.append((worker, task))

        cost = total_cost(assignment, cost_matrix)

        if cost < best_cost:
            best_cost = cost
            best_assignment = assignment

    return best_assignment, best_cost


print("\n========== ASSIGNMENT PROBLEM Q13 - TEST CASE 1 ==========")

cost_matrix1 = [
    [3, 10, 7],
    [8, 5, 12],
    [4, 6, 9]
]

assignment, cost = assignment_problem(cost_matrix1)

print("Optimal Assignment:")

for worker, task in assignment:
    print(
        f"(worker {worker + 1}, task {task + 1})"
    )

print("Total Cost:", cost)


print("\n========== ASSIGNMENT PROBLEM Q13 - TEST CASE 2 ==========")

cost_matrix2 = [
    [15, 9, 4],
    [8, 7, 18],
    [6, 12, 11]
]

assignment, cost = assignment_problem(cost_matrix2)

print("Optimal Assignment:")

for worker, task in assignment:
    print(
        f"(worker {worker + 1}, task {task + 1})"
    )

print("Total Cost:", cost)
```

#### QUESTION 14: EXHAUSTIVE SEARCH - 0/1 KNAPSACK

```python
def total_value(items, values):
    total = 0

    for item in items:
        total += values[item]

    return total


def is_feasible(items, weights, capacity):
    total_weight = 0

    for item in items:
        total_weight += weights[item]

    return total_weight <= capacity


def knapsack_exhaustive(weights, values, capacity):
    n = len(weights)

    best_items = []
    best_value = 0

    for r in range(n + 1):

        for subset in combinations(range(n), r):

            if is_feasible(subset, weights, capacity):

                value = total_value(subset, values)

                if value > best_value:
                    best_value = value
                    best_items = list(subset)

    return best_items, best_value


print("\n========== KNAPSACK Q14 - TEST CASE 1 ==========")

weights1 = [2, 3, 1]
values1 = [4, 5, 3]
capacity1 = 4

items, value = knapsack_exhaustive(
    weights1,
    values1,
    capacity1
)

print("Optimal Selection:", items)
print("Total Value:", value)


print("\n========== KNAPSACK Q14 - TEST CASE 2 ==========")

weights2 = [1, 2, 3, 4]
values2 = [2, 4, 6, 3]
capacity2 = 6

items, value = knapsack_exhaustive(
    weights2,
    values2,
    capacity2
)

print("Optimal Selection:", items)
print("Total Value:", value)
```

#### QUESTION 15: SUBSET SUM - ALL POSSIBLE SUBSETS

```python
def all_subset_sum(numbers, target):
    n = len(numbers)
    found = []

    for r in range(n + 1):

        for subset in combinations(numbers, r):

            if sum(subset) == target:
                found.append(list(subset))

    return found


print("\n========== SUBSET SUM Q15 ==========")

numbers = [15, 10, 12, 7, 5]
target = 22

subsets = all_subset_sum(numbers, target)

if subsets:

    for subset in subsets:
        print("Subset found:", subset)
        print("Sum:", sum(subset))

else:
    print("No subset found")
```
